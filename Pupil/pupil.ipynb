```python
import zmq
import json
import time

ctx = zmq.Context()
pupil_remote = zmq.Socket(ctx, zmq.REQ)
pupil_remote.connect('tcp://127.0.0.1:50020')
```

```python
# start recording
pupil_remote.send_string('R')
print(pupil_remote.recv_string())

time.sleep(5)
pupil_remote.send_string('r')
print(pupil_remote.recv_string())
```

```python
import zmq
import json

ctx = zmq.Context()
socket = ctx.socket(zmq.REQ)
socket.connect("tcp://localhost:50020")  # or use IP if remote

def pupil_command(cmd, payload=None):
    msg = {"subject": cmd, "mode": "push"}
    if payload:
        msg["payload"] = payload
    socket.send_string(json.dumps(msg))
    return socket.recv_string()

pupil_command("R")
```
